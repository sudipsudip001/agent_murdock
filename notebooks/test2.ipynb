```python
answer = await graph.run(
    question="Will Cole Palmer play in the FIFA World Cup 2026?",
    thread_id="user-123-session-1",
)
print(answer)

answer = await graph.run(
    question="What club does he play for?",
    thread_id="user-123-session-1",
)
print(answer)

answer = await graph.run(
    question="How many goals and assists did he have in this premier league season?",
    thread_id="user-123-session-1",
)
print(answer)

await graph.teardown()
```
